```python
import random
```

```python
names = open('names.txt').read().splitlines()
names = [n.strip().lower() for n in names if n.strip()]
print(f"Loaded {len(names)} names, e.g. {names[:5]}")
```

```python
counts = {}
for name in names:
    chars = ['<S>'] + list(name) + ['<E>']
    for c1, c2 in zip(chars, chars[1:]):
        if c1 not in counts:
            counts[c1] = {}
        counts[c1][c2] = counts[c1].get(c2, 0) + 1
```

```python
print("\nMost common first letters:")
print(sorted(counts['<S>'].items(), key=lambda x: -x[1])[:5])
print("\nMost common letters after 'q':")
print(sorted(counts['q'].items(), key=lambda x: -x[1])[:5])
```

```python
print("\nGenerated names:")
random.seed(4256)
for _ in range(10):
    ch = '<S>'
    out = []
    while True:
        next_chars = list(counts[ch].keys())
        weights = list(counts[ch].values())
        ch = random.choices(next_chars, weights=weights)[0]
        if ch == '<E>':
            break
        out.append(ch)
    print(''.join(out))
```
